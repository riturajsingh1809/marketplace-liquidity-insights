```python
import pandas as pd
import numpy as np

# Load cleaned dataset
df = pd.read_excel("Cleaned_Marketplace_Liquidity.xlsx")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

# Convert date column
df["Event_Date"] = pd.to_datetime(df["Event_Date"], errors="coerce")

# Create date features
df["Month"] = df["Event_Date"].dt.month
df["Month_Name"] = df["Event_Date"].dt.strftime("%B")
df["Year"] = df["Event_Date"].dt.year

# --------------------------------------------------
# 1. Overall Business Summary
# --------------------------------------------------

total_events = len(df)
total_companies = df["Company_ID"].nunique()
total_students = df["Student_ID"].nunique()
total_jobs = df["Job_ID"].nunique()

company_posts = (df["Event_Type"] == "Company Posted").sum()
student_applications = (df["Event_Type"] == "Student Applied").sum()
company_shortlists = (df["Event_Type"] == "Company Shortlisted").sum()

overall_summary = pd.DataFrame({
    "Metric": [
        "Total Events",
        "Unique Companies",
        "Unique Students",
        "Unique Jobs",
        "Company Posts",
        "Student Applications",
        "Company Shortlists"
    ],
    "Value": [
        total_events,
        total_companies,
        total_students,
        total_jobs,
        company_posts,
        student_applications,
        company_shortlists
    ]
})

# --------------------------------------------------
# 2. Marketplace Funnel Analysis
# --------------------------------------------------

funnel = pd.DataFrame({
    "Stage": [
        "Company Posted",
        "Student Applied",
        "Company Shortlisted"
    ],
    "Events": [
        company_posts,
        student_applications,
        company_shortlists
    ]
})

if company_posts > 0:
    application_rate = (student_applications / company_posts) * 100
else:
    application_rate = 0

if student_applications > 0:
    shortlist_rate = (company_shortlists / student_applications) * 100
else:
    shortlist_rate = 0

if company_posts > 0:
    overall_conversion = (company_shortlists / company_posts) * 100
else:
    overall_conversion = 0

funnel_metrics = pd.DataFrame({
    "Metric": [
        "Application Rate",
        "Shortlist Rate",
        "Overall Shortlist Conversion"
    ],
    "Percentage": [
        round(application_rate, 2),
        round(shortlist_rate, 2),
        round(overall_conversion, 2)
    ]
})

# --------------------------------------------------
# 3. Event Type Analysis
# --------------------------------------------------

event_analysis = (
    df.groupby("Event_Type")
    .agg(
        Total_Events=("Event_ID", "count"),
        Unique_Companies=("Company_ID", "nunique"),
        Unique_Students=("Student_ID", "nunique"),
        Unique_Jobs=("Job_ID", "nunique"),
        Average_Search_Time=("Search_Time_Seconds", "mean"),
        Average_Processing_Time=("Processing_Time_Hours", "mean")
    )
    .reset_index()
)

event_analysis["Average_Search_Time"] = event_analysis[
    "Average_Search_Time"
].round(2)

event_analysis["Average_Processing_Time"] = event_analysis[
    "Average_Processing_Time"
].round(2)

# --------------------------------------------------
# 4. Company Performance Analysis
# --------------------------------------------------

company_analysis = (
    df.groupby("Company_ID")
    .agg(
        Total_Events=("Event_ID", "count"),
        Jobs=("Job_ID", "nunique"),
        Applications=("Student_ID", "nunique"),
        Average_Processing_Time=("Processing_Time_Hours", "mean"),
        Average_Search_Time=("Search_Time_Seconds", "mean")
    )
    .reset_index()
)

company_analysis["Average_Processing_Time"] = company_analysis[
    "Average_Processing_Time"
].round(2)

company_analysis["Average_Search_Time"] = company_analysis[
    "Average_Search_Time"
].round(2)

company_analysis = company_analysis.sort_values(
    "Total_Events",
    ascending=False
)

# --------------------------------------------------
# 5. Student Engagement Analysis
# --------------------------------------------------

student_analysis = (
    df.groupby("Student_ID")
    .agg(
        Total_Events=("Event_ID", "count"),
        Companies_Interacted=("Company_ID", "nunique"),
        Jobs_Interacted=("Job_ID", "nunique"),
        Average_Search_Time=("Search_Time_Seconds", "mean")
    )
    .reset_index()
)

student_analysis["Average_Search_Time"] = student_analysis[
    "Average_Search_Time"
].round(2)

student_analysis = student_analysis.sort_values(
    "Total_Events",
    ascending=False
)

# --------------------------------------------------
# 6. Skill Analysis
# --------------------------------------------------

skill_analysis = (
    df.groupby("Skill")
    .agg(
        Total_Events=("Event_ID", "count"),
        Unique_Companies=("Company_ID", "nunique"),
        Unique_Students=("Student_ID", "nunique"),
        Applications=("Event_ID", lambda x: (df.loc[x.index, "Event_Type"] == "Student Applied").sum()),
        Shortlists=("Event_ID", lambda x: (df.loc[x.index, "Event_Type"] == "Company Shortlisted").sum())
    )
    .reset_index()
)

skill_analysis["Shortlist_Rate"] = np.where(
    skill_analysis["Applications"] > 0,
    (skill_analysis["Shortlists"] / skill_analysis["Applications"]) * 100,
    0
)

skill_analysis["Shortlist_Rate"] = skill_analysis[
    "Shortlist_Rate"
].round(2)

skill_analysis = skill_analysis.sort_values(
    "Total_Events",
    ascending=False
)

# --------------------------------------------------
# 7. Region Analysis
# --------------------------------------------------

region_analysis = (
    df.groupby("Region")
    .agg(
        Total_Events=("Event_ID", "count"),
        Companies=("Company_ID", "nunique"),
        Students=("Student_ID", "nunique"),
        Jobs=("Job_ID", "nunique"),
        Average_Search_Time=("Search_Time_Seconds", "mean"),
        Average_Processing_Time=("Processing_Time_Hours", "mean")
    )
    .reset_index()
)

region_analysis["Average_Search_Time"] = region_analysis[
    "Average_Search_Time"
].round(2)

region_analysis["Average_Processing_Time"] = region_analysis[
    "Average_Processing_Time"
].round(2)

region_analysis = region_analysis.sort_values(
    "Total_Events",
    ascending=False
)

# --------------------------------------------------
# 8. Application Status Analysis
# --------------------------------------------------

status_analysis = (
    df.groupby("Application_Status")
    .agg(
        Total_Events=("Event_ID", "count"),
        Unique_Students=("Student_ID", "nunique"),
        Unique_Companies=("Company_ID", "nunique")
    )
    .reset_index()
)

status_analysis["Percentage"] = (
    status_analysis["Total_Events"] /
    status_analysis["Total_Events"].sum() * 100
).round(2)

# --------------------------------------------------
# 9. Monthly Trend Analysis
# --------------------------------------------------

monthly_analysis = (
    df.groupby(["Year", "Month", "Month_Name"])
    .agg(
        Total_Events=("Event_ID", "count"),
        Company_Posts=("Event_ID", lambda x: (df.loc[x.index, "Event_Type"] == "Company Posted").sum()),
        Student_Applications=("Event_ID", lambda x: (df.loc[x.index, "Event_Type"] == "Student Applied").sum()),
        Company_Shortlists=("Event_ID", lambda x: (df.loc[x.index, "Event_Type"] == "Company Shortlisted").sum()),
        Unique_Companies=("Company_ID", "nunique"),
        Unique_Students=("Student_ID", "nunique")
    )
    .reset_index()
)

monthly_analysis["Application_Rate"] = np.where(
    monthly_analysis["Company_Posts"] > 0,
    monthly_analysis["Student_Applications"] /
    monthly_analysis["Company_Posts"] * 100,
    0
)

monthly_analysis["Shortlist_Rate"] = np.where(
    monthly_analysis["Student_Applications"] > 0,
    monthly_analysis["Company_Shortlists"] /
    monthly_analysis["Student_Applications"] * 100,
    0
)

monthly_analysis["Application_Rate"] = monthly_analysis[
    "Application_Rate"
].round(2)

monthly_analysis["Shortlist_Rate"] = monthly_analysis[
    "Shortlist_Rate"
].round(2)

# --------------------------------------------------
# 10. Search Performance Analysis
# --------------------------------------------------

search_analysis = pd.DataFrame({
    "Metric": [
        "Average Search Time",
        "Minimum Search Time",
        "Maximum Search Time",
        "Median Search Time"
    ],
    "Seconds": [
        round(df["Search_Time_Seconds"].mean(), 2),
        round(df["Search_Time_Seconds"].min(), 2),
        round(df["Search_Time_Seconds"].max(), 2),
        round(df["Search_Time_Seconds"].median(), 2)
    ]
})

# --------------------------------------------------
# 11. Processing Performance Analysis
# --------------------------------------------------

processing_analysis = pd.DataFrame({
    "Metric": [
        "Average Processing Time",
        "Minimum Processing Time",
        "Maximum Processing Time",
        "Median Processing Time"
    ],
    "Hours": [
        round(df["Processing_Time_Hours"].mean(), 2),
        round(df["Processing_Time_Hours"].min(), 2),
        round(df["Processing_Time_Hours"].max(), 2),
        round(df["Processing_Time_Hours"].median(), 2)
    ]
})

# --------------------------------------------------
# 12. Data Quality Check
# --------------------------------------------------

data_quality = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Missing Values",
        "Unique Events",
        "Unique Companies",
        "Unique Students",
        "Unique Jobs"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        df["Event_ID"].nunique(),
        df["Company_ID"].nunique(),
        df["Student_ID"].nunique(),
        df["Job_ID"].nunique()
    ]
})

# --------------------------------------------------
# 13. Key Business Insights
# --------------------------------------------------

top_skill = (
    skill_analysis.iloc[0]["Skill"]
    if len(skill_analysis) > 0 else "N/A"
)

top_region = (
    region_analysis.iloc[0]["Region"]
    if len(region_analysis) > 0 else "N/A"
)

top_company = (
    company_analysis.iloc[0]["Company_ID"]
    if len(company_analysis) > 0 else "N/A"
)

insights = pd.DataFrame({
    "Business Insight": [
        "Total marketplace events",
        "Most active skill",
        "Most active region",
        "Most active company",
        "Application rate",
        "Shortlist rate",
        "Overall shortlist conversion",
        "Average search time",
        "Average processing time"
    ],
    "Value": [
        total_events,
        top_skill,
        top_region,
        top_company,
        round(application_rate, 2),
        round(shortlist_rate, 2),
        round(overall_conversion, 2),
        round(df["Search_Time_Seconds"].mean(), 2),
        round(df["Processing_Time_Hours"].mean(), 2)
    ]
})

# --------------------------------------------------
# 14. Save All Analysis Results in ONE File
# --------------------------------------------------

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:

    overall_summary.to_excel(
        writer,
        sheet_name="Overall Summary",
        index=False
    )

    funnel.to_excel(
        writer,
        sheet_name="Funnel Analysis",
        index=False
    )

    funnel_metrics.to_excel(
        writer,
        sheet_name="Funnel Metrics",
        index=False
    )

    event_analysis.to_excel(
        writer,
        sheet_name="Event Analysis",
        index=False
    )

    company_analysis.to_excel(
        writer,
        sheet_name="Company Analysis",
        index=False
    )

    student_analysis.to_excel(
        writer,
        sheet_name="Student Analysis",
        index=False
    )

    skill_analysis.to_excel(
        writer,
        sheet_name="Skill Analysis",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    status_analysis.to_excel(
        writer,
        sheet_name="Status Analysis",
        index=False
    )

    monthly_analysis.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    search_analysis.to_excel(
        writer,
        sheet_name="Search Performance",
        index=False
    )

    processing_analysis.to_excel(
        writer,
        sheet_name="Processing Performance",
        index=False
    )

    data_quality.to_excel(
        writer,
        sheet_name="Data Quality",
        index=False
    )

    insights.to_excel(
        writer,
        sheet_name="Business Insights",
        index=False
    )

print("Python analysis completed successfully.")
print(f"Final output created: {output_file}")
```
